```python
import os
import polars as pl

# Resolve path to your Gold Delta analytics table
GOLD_PATH = "/Users/intuitivecode/Code/deltastreams/data/tables/gold_alerts"

# Load the ML-classified Gold layer natively
df_gold = pl.read_delta(GOLD_PATH)

print("┌────────────────────────────────────────────────────────┐")
print("│             GOLD INTELLIGENCE SUMMARY           │")
print("└────────────────────────────────────────────────────────┘")
print(f"Total Transactions Audited : {df_gold.height}")
print(f"Total Anomalies Flagged     : {df_gold.filter(pl.col('anomaly_score') == -1).height}")
print(f"Pipeline Infrastructure Cost: $0.00 (Bare-Metal Local Run)\n")

print(" --- DETECTED CYBER-FINANCE MATRIX ---")
summary = (df_gold
    .group_by("market_regime")
    .agg(pl.len().alias("Captured_Events"))
    .sort("Captured_Events", descending=True)
)
print(summary)

print("\n --- SAMPLE DRILLDOWN: RECENT FLAGGED ANOMALIES ---")
print(df_gold
    .filter(pl.col("anomaly_score") == -1)
    .sort("_ingested_at", descending=True)
    .select([
        "product_id", 
        "price", 
        "size", 
        "price_velocity", 
        "size_velocity_ratio", 
        "market_regime"
    ])
    .head(5)
)
```
